## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
import pandas as pd

# Load dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["action_score"] = (
    (df["impressions_90d"] * 0.4) +
    (df["clicks_90d"] * 0.3) +
    (df["pageviews_90d"] * 0.2) +
    (df["sessions_90d"] * 0.1)
)

ranked = df.sort_values("action_score",ascending=False).copy()

# Reason codes
def reason(row):
    if row["impressions_90d"] > 10000:
        return "HIGH_VISIBILITY"
    elif row["clicks_90d"] < 10:
        return "LOW_CTR"
    elif row["days_since_last_update"] > 180:
        return "STALE_CONTENT"
    else:
        return "MODERATE_PRIORITY"

ranked["reason_code"] = ranked.apply(reason, axis=1)

# Action labels
def action(row):
    if row["reason_code"] == "STALE_CONTENT":
        return "REFRESH_CONTENT"
    elif row["reason_code"] == "LOW_CTR":
        return "OPTIMIZE_TITLE"
    elif row["reason_code"] == "HIGH_VISIBILITY":
        return "HIGH_PRIORITY_REVIEW"
    else:
        return "MONITOR"

ranked["action_label"] = ranked.apply(action, axis=1)

queue = ranked[
    [
        "content_id",
        "action_score",
        "reason_code",
        "action_label"
    ]
].head(20)

queue
```

```python
print("""
Ranked Actions

Pages are ranked using a transparent action score based on historical visibility and engagement metrics.

Reason codes explain why a page appears in the queue.

The queue is intended to support human review rather than automate publishing decisions.
""")
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

```python
print("""
Intended Use

This content action playbook is designed to help content teams prioritize pages for manual review.

The ranked queue highlights pages that show strong historical visibility or engagement and may benefit from actions such as content refresh, title optimization or further investigation.

The recommendations are intended for decision-support and should not be interpreted as automatic publishing instructions.

Limits

The recommendations are based only on historical search and engagement metrics available in this dataset.

The model does not evaluate content quality, factual accuracy, user intent, seasonality or future search engine updates.

Therefore, every recommendation should be reviewed by a human editor before implementation.
""")
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

```python
print("""
Human Review

Before taking action on any recommendation, a reviewer should verify:

1. Whether the content is still accurate and relevant.
2. Whether search intent has changed.
3. Whether the page has already been updated recently.
4. Whether seasonal or temporary trends explain the observed performance.
5. Whether improving the page aligns with business priorities.

No-Go List

The system should NOT automatically:

- Publish or modify content.
- Delete pages.
- Change titles without editorial approval.
- Make decisions based only on model predictions.
- Override human judgment for important business decisions.

The ranked queue should always be treated as decision-support rather than an automated workflow.
""")
```
