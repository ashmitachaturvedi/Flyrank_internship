## 1. Question

*The research question and the decision it supports.*

```python
print("""
Research Question

Can historical search and engagement metrics be used to identify content pages that are more likely to experience declining search performance?

Decision Supported

The goal of this work is to support content teams in prioritizing pages for manual review and content refresh.

Rather than replacing editorial judgment, the model provides a ranked list of pages that may deserve attention based on observed historical performance.

The recommendations are intended for decision-support and should always be reviewed by a human before implementation.
""")
```

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

```python
print("""
Data

This project was developed using the FlyRank ML Internship dataset built from large-scale production search and engagement data.

The analysis was performed on the anonymized content refresh dataset containing approximately 30,000 content records used for the internship exercises.

The dataset includes historical search performance, engagement metrics, content characteristics and content freshness indicators.

Examples of features include:

- Search impressions
- Clicks
- Pageviews
- Sessions
- Content age
- Average search position
- Click-through rate (CTR)
- Engagement rate

The target label was derived from the observed trend direction of each content page.

Private client names, URLs, search queries and other sensitive information were excluded throughout the analysis to maintain public-safe reporting.
""")
```

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

```python
print("""
Methodology

The objective of this study was to identify content pages that are more likely to experience declining search performance using historical search and engagement metrics.

Target Label

The target variable was created from the observed trend_direction field, where pages labeled "down" were assigned the positive class.

Features

The model used historical search performance, engagement metrics, content characteristics and freshness indicators, including impressions, clicks, pageviews, sessions, content age, average position, click-through rate and engagement rate.

Baseline

A simple baseline was established before training machine learning models. Logistic Regression and Random Forest were then evaluated using the same prediction task.

Validation Design

The final evaluation used a grouped validation split based on client_id so that pages from the same client did not appear in both the training and testing sets. This provides a more realistic estimate of performance on previously unseen clients.

Leakage Checks

Potential leakage was reviewed before model training. The trend_direction field was used only as the target label, trend_pct was excluded from the feature set, and no product-generated decision outputs were used as model inputs.
""")
```

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

```python
import pandas as pd

results = pd.DataFrame({
    "Validation": [
        "Random Split",
        "Grouped Split"
    ],
    "Model": [
        "Random Forest",
        "Random Forest"
    ],
    "Accuracy": [
        0.9098,
        0.8861
    ],
    "Precision": [
        0.8981,
        0.8766
    ],
    "Recall": [
        0.9403,
        0.9044
    ],
    "F1 Score": [
        0.9187,
        0.8903
    ]
})

print("Model Performance Summary")
display(results)

print("""
Results Summary

The Random Forest model achieved the strongest measured performance among the evaluated models.

When evaluated using a grouped validation split, the performance decreased slightly compared with the random split.

This result suggests that the grouped split provides a more realistic estimate of model performance on previously unseen clients.

The grouped validation results were used as the primary evaluation because they better reflect real-world deployment.
""")
```

## 5. Limitations

*What this work cannot claim.*

```python
print("""
Limitations

This work is intended as a decision-support tool rather than an automated decision system.

The model was trained and evaluated using historical search and engagement data available in the internship dataset.

The results should not be interpreted as evidence that any feature causes improvements or declines in search performance.

The model does not account for factors such as future search engine algorithm updates, content quality, user intent, seasonal effects or external business changes.

Although grouped validation provides a more realistic estimate of performance, the reported metrics may differ when applied to new datasets or production environments.

Therefore, the ranked recommendations should always be reviewed by human editors before any content changes are made.
""")
```
